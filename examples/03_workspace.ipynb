
#### Dataset versioning

After ingesting your dataset through the pipeline, you can manage and version it with DVC.

DVC works alongside Git. It stores large data files in local or remote storage, such as a local directory or cloud bucket. DVC then creates small metadata files that reference the stored data. You can commit these metadata files to Git and push them to a repository such as GitHub, making datasets easier to track, share, and reproduce.

````python
# on your workspace folder do

!pip install dvc

!dvc init

# This creates data.dvc and adds data/ to .gitignore. Git will track data.dvc, but not the large dataset itself.
!dvc add data/

# add your remote storage location, DO NOT RUN them all, choose one
!dvc remote add -d storage /mnt/shared/geosave-dvc
!dvc remote add -d storage s3://my-bucket/geosave-data
!dvc remote modify storage endpointurl https://s3.example.com
!dvc remote add -d storage gdrive://FOLDER_ID/geosave-data

# push the data to the remote storage to update the data
!dvc push

# or you can push it to huggingface
from huggingface_hub import upload_folder, upload_file

# or best practice to to upload to bucket storage

# then you are save to upload to remote git repository and share the experiment with your team.
!git add data.dvc .gitignore
!git commit -m "Add data to DVC"
!git push origin main

# If you want to share the experiment with your team, they can clone the repository and run
!dvc pull
````
